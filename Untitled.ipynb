```python
import pandas as pd
import numpy as np
```

```python
CSV_PATH = "StudentsPerformance.csv"
```

```python
try:
    df = pd.read_csv(CSV_PATH)
    print(f"Loaded real dataset from {CSV_PATH}\n")
except FileNotFoundError:
    print(f"Couldn't find {CSV_PATH} -- using a small sample dataset instead.\n")
    # Same column structure as the real Kaggle dataset, with a few missing
    # values (np.nan) mixed in on purpose, so cleaning has something to do.
    data = {
        "gender": ["female", "male", "female", "male", "female", "male", "female", "male"],
        "race/ethnicity": ["group B", "group C", "group A", "group B", "group C",
                            "group A", "group D", "group B"],
        "parental level of education": ["bachelor's degree", "some college",
                                         "master's degree", "associate's degree",
                                         "high school", "some college",
                                         "bachelor's degree", "high school"],
        "test preparation course": ["none", "completed", "none", "none",
                                     "completed", "none", "completed", "none"],
        "math score": [72, 69, np.nan, 47, 76, 71, np.nan, 88],
        "reading score": [72, 90, 95, 57, 78, np.nan, 79, 64],
        "writing score": [74, 88, 93, 44, np.nan, 78, 80, 60],
    }
    df = pd.DataFrame(data)
```

```python
print("--- First 5 rows ---")
print(df.head())
```

```python
print("\n--- Shape (rows, columns) ---")
print(df.shape)
```

```python
print("\n--- Missing values per column ---")
print(df.isnull().sum())
```

```python
score_cols = ["math score", "reading score", "writing score"]
 
for col in score_cols:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].mean())
 
print("\n--- Missing values after cleaning ---")
print(df.isnull().sum())
```

```python
print("\n--- Score statistics ---")
for col in score_cols:
    if col in df.columns:
        avg = df[col].mean()
        mx = df[col].max()
        mn = df[col].min()
        print(f"{col}: average = {avg:.2f}, max = {mx}, min = {mn}")
```

```python
print("\n--- Summary table (pandas describe) ---")
print(df[score_cols].describe().loc[["mean", "max", "min"]])
```

```python
df["overall average"] = df[score_cols].mean(axis=1)
 
print("\n--- Final cleaned dataset with overall average ---")
print(df.head())
```

```python
print("\nQ1: Create a DataFrame manually")
manual_df = pd.DataFrame({
    "name": ["Asha", "Ravi", "Meera"],
    "marks": [85, np.nan, 90]
})
print(manual_df)
```

```python
print("\nQ2: Handle missing values with fillna()")
manual_df["marks"] = manual_df["marks"].fillna(manual_df["marks"].mean())
print(manual_df)
```

```python
print("\nQ3: Sort by 'marks' column (descending)")
sorted_df = manual_df.sort_values(by="marks", ascending=False)
print(sorted_df)
```

```python
print("\nQ4: Mean, median, mode of 'marks'")
print("Mean:", manual_df["marks"].mean())
print("Median:", manual_df["marks"].median())
print("Mode:", manual_df["marks"].mode()[0])
```
